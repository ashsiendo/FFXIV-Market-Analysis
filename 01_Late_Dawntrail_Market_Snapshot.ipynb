# **FFXIV Market Dynamics - Study 1**

## Late-Dawntrail Selected-World Market Snapshot

> **How do selected FFXIV NA worlds differ in price, sales activity, available supply, and seller competition during the October-centered late-Dawntrail observation period?**

The snapshot examines Goblin, Jenova, Adamantoise, and Famfrit using FFXIV Market Board data provided by Universalis.

The study grew out of exploring different FFXIV worlds and becoming curious about how their MB economies compare; it will implement a predefined list of items, collect current listings across the four worlds within a closely timed window, and review a standard period of past transactions. Goblin is used as a personal reference world, but is not representative of all NA worlds.

**Right now, the project is testing the API and setting up the research methods. The final list of items and collection rules will be decided before primary data collection.**
